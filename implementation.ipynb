```python
from falcon import Falcon

# Create Falcon with a valid preset
model = Falcon(256)
print(f"Model created with preset 8: {model}")

sk, vk = model.keygen()
print(f"Secret key: {sk}")
print(f"Verification key: {vk}")

message = "Hello, Falcon!"
bmessage = message.encode('utf-8')
sig = model.sign(sk, bmessage)
print(f"Signature: {sig}")
is_valid = model.verify(vk, bmessage, sig)
print(f"Signature valid: {is_valid}")
```
