```python
from falcon import Falcon

falcon = Falcon(512)
```

```python
secret_key, public_key = falcon.keygen()

print("Public key length:", len(public_key))
print("Secret key length:", len(secret_key))
```

```python
message = b"TEST COMMAND"

signature = falcon.sign(secret_key, message)

print("Signature length:", len(signature))
```

```python
try:
    valid = falcon.verify(public_key, message, signature)
    if valid:
        print("VALID SIGNATURE → PASS")
    else:
        print("VERIFICATION FAILED → FAIL (returned False)")
except Exception as e:
    print("VERIFICATION FAILED → FAIL", e)
```

```python
tampered_message = b"TEST COMMAND MODIFIED"

signature1 = falcon.sign(secret_key, tampered_message)

print("Tampered Signature length:", len(signature1))
```
